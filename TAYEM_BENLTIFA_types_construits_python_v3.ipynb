##<font color="purple">1. Introduction</font>

##<font color="purple">2. P-uplets</font>

###<font color="red">Définition</font>

Un objet de type tuple, un p-uplet, est une suite ordonnée d’éléments qui peuvent être chacun de n’importe quel type. On parlera indifféremment de p-uplet ou de tuple.

**Création d'un p-uplet**

Pour créer un p-uplet non vide, on écrit $n$ valeurs séparées par des virgules. Par exemple :

```
t1=   ("a", "b", "c", 3) #pour un tuple à 4 éléments
t2=   "a", #pour un tuple à 1 élément, la virgule suivant l'élément est obligatoire
t3=() #pour un tuple à 0 éléments
t4=tuple() # autre façon de créer un tuple vide
```

Pour écrire un p-uplet qui contient un n-uplet, l’utilisation de parenthèses est nécessaire. Voici un exemple avec un tuple à 2 éléments dont le second est un tuple : $\texttt{t = 3, ("a","b","c")}$. En général, les parenthèses sont obligatoires dès que l’écriture d’un p-uplet est contenue dans une expression plus longue. Dans tous les cas, les parenthèses peuvent améliorer la lisibilité.

**Opérations**

Prenons  pour  exemple  l’affectation $\texttt{a, b, c = 1, 2, 3}$.  Ceci  signifie  que  le  tuple $\texttt{(a, b, c)}$ prend  pour  valeur  le  tuple$\texttt{(1, 2, 3)}$, autrement dit, les valeurs respectives des variables $\texttt{a}$, $\texttt{b}$ et $\texttt{c}$ sont $1$, $2$ et $3$. En particulier, l’instruction $\texttt{a, b = b, a }$ permet d’échanger les valeurs des deux variables $\texttt{a}$ et $\texttt{b}$. Les valeurs des éléments d’un tuple peuvent ainsi être stockées dans des variables.

```
t = 1, 2, 3
a, b, c = t
```

```
b
```

Cette syntaxe s’utilise souvent avec une fonction qui renvoie un tuple. Voici  un  exemple  avec  une  fonction  qui  calcule  et  renvoie  les  longueurs  des  trois  côtés  d’un  triangle  ABC.  La  fonction  prend  en paramètres trois p-uplets représentant les coordonnées des trois points. On importe au préalable la fonction racine carrée $\texttt{sqrt}$ du module $\texttt{math}$.

```
from math import sqrt
def longueurs(A,B, C):
  xA,yA = A
  xB,yB = B
  xC,yC = C
  dAB= sqrt((xB - xA) ** 2 + (yB - yA) ** 2)
  dBC= sqrt((xC - xB) ** 2 + (yC - yB) ** 2)
  dAC= sqrt((xC - xA) ** 2 + (yC - yA) ** 2)
  return dAB, dBC, dAC
```

La fonction étant définie, nous l’utilisons dans l’interpréteur :

```
M = (3.4, 7.8)
N = (5, 1.6)
P = (-3.8, 4.3)
dMN, dNP, dMP = longueurs(M, N, P)
dMP
```

```
*_,dMP=longueurs(M,N,P)
dMP
```

```
dMP=longueurs(M,N,P)[-1]
dMP
```

Dans cet exemple, une même liste a deux noms ($\texttt{liste1}$ et $\texttt{liste2}$) et $\texttt{liste2}$ n’est pas une copie de $\texttt{liste1}$. Pour obtenir une copie, il faut créer une nouvelle liste.

Par exemple :

```
%%tutor

liste1=["a","b","c"]
liste2 =list(liste1)
```

Il s’agit d’une copie superficielle, disons de niveau 1. Pour comprendre le fonctionnement, il faut considérer qu’une liste est juste une adresse (en mémoire) qui contient les adresses de ses éléments. Avec cette copie superficielle, une nouvelle liste est créée avec une nouvelle adresse. Les éléments gardent eux la même adresse.

Examinons  un  autre  exemple  avec $\texttt{liste = 2 * [["a","b"]]}$.  Cela  revient  à  écrire $\texttt{liste1 = ["a","b"]}$, puis $\texttt{liste = 2 * [liste1]}$.  Le  résultat  est $\texttt{[['a','b'], ['a','b']]}$.  Les  deux  éléments  ont  ici  la  même  adresse, celle de $\texttt{liste1}$. Donc une modification d’un élément de $\texttt{liste1}$ concerne les deux éléments de $\texttt{liste}$. Que ce soit $\texttt{liste1[1] ="c"}$ ou $\texttt{liste[0][1] ="c"}$, le résultat est le même, $\texttt{[['a','c'], ['a','c']]}$.

```
%%tutor

liste=2*[["a","b"]]
liste[0][1]="c"
print(liste)
```

Par contre avec l’instruction $\texttt{liste = [["a","b"]for i in range(2)]}$, une première liste $\texttt{["a","b"]}$ est créée quand $\texttt{i}$ prend  la  valeur $\texttt{0}$,  puis  une  seconde  liste  quand $\texttt{i}$  prend  la  valeur  $\texttt{1}$.  Ces  deux  listes  sont  distinctes,  (les  adresses  sont différentes), donc modifier l’une n’a aucune conséquence sur l’autre.

```
%%tutor

liste = [["a","b"]for i in range(2)]
liste

```

Considérons le cas d’une liste dont les éléments sont des listes. Une copie superficielle comme ci-dessus crée une nouvelle liste avec une nouvelle adresse et les éléments de cette copie ont eux la même adresse que les éléments de la liste initiale. Par exemple avec $\texttt{liste1 = [['a','b'], ['c','d']]}$ puis $\texttt{liste2 =list(liste1)}$, les adresses de $\texttt{liste1}$ et $\texttt{liste2}$ sont distinctes mais les adresses de $\texttt{liste1[0]}$ et $\texttt{liste2[0]}$ sont identiques. Donc la modification d’un élément de $\texttt{liste1[0]}$ se  répercute  sur $\texttt{liste2[0]}$.

```
%%tutor

liste1 = [['a','b'], ['c','d']]
liste2 =list(liste1)
```

Pour  obtenir  une  copie "en  profondeur",  passer  au  niveau  2,  il  faut  donner  de  nouvelles adresses pour les listes éléments d’une liste.

Des fonctions de copie sont disponibles à partir du module $\texttt{copy}$. En particulier, dans le cas d’une liste de listes, pour effectuer une copie en profondeur, nous disposons de la fonction $\texttt{deepcopy}$.

```
%%tutor

from copy import deepcopy
liste1 = [["a","b"], ["c","d"]]
liste2 = deepcopy(liste1)
liste2[1][0] ="e"
print(liste1)
print(liste2)
```

##<font color="purple">4. Dictionnaires</font>

###<font color="red">Définition</font>
Les éléments d’une liste sont repérés par des indices 0, 1, 2, ... Une différence essentielle avec un dictionnaire, objet de type $\texttt{dict}$, est que les indices sont remplacés par des objets du type $\texttt{str}$, $\texttt{float}$, $\texttt{tuple}$. On les appelle des clés et à chaque clé correspond une valeur. Ces clés ne sont pas ordonnées.

###<font color="red">Création</font>
Les éléments d’un dictionnaire sont des couples clé-valeur. Un dictionnaire est créé avec des accolades, les différents couples étant séparés par des virgules. La clé et la valeur correspondante d’un élément sont séparées par deux-points.

Exemple : $\texttt{dico = {"A": 0,"B": 1,"C": 2,"D": 3}}$.

On peut construire un dictionnaire en compréhension comme avec les listes :


```
d = {chr(65+i): i for i in range(26)}
d
```

On peut convertir une liste de listes à deux éléments en dictionnaire avec la fonction $\texttt{dict}$.

```
liste = [['A', 0], ['B', 1], ['C', 2]]
d =dict(liste)
d
```

###<font color="red">Utilisation</font>
**Accès aux éléments**

Pour accéder aux clés ou aux valeurs, nous avons les méthodes $\texttt{keys}$ et $\texttt{values}$.

```
d = {'A': 0,'B': 1,'C': 2}
```

```
d.keys()
```

```
d.values()
```

Pour l’accès à l’ensemble des couples clés-valeurs, nous utilisons la méthode $\texttt{items}$.

```
d.items()
```

Remarque: les couples clés-valeurs obtenus sont du type $\texttt{tuple}$.

Nous pouvons tester l’appartenance à un dictionnaire avec le mot clé $\texttt{in}$.

```
"A" in d  # teste si "A" est une clé
```

```
3 in d.values()# teste si 3 est une valeur
```

```
('C', 2) in d.items()# teste si ('C', 2) est un couple clé-valeur
```

On en déduit trois manières de parcourir un dictionnaire. Il est possible d’utiliser les clés, les valeurs, ou les couples clés-valeurs.

Voici un exemple avec les clés :

```
for key in d:
  print(key)
```

L’accès à une valeur particulière s’obtient en précisant la clé. Par exemple, $\texttt{d["A"]}$ a la valeur $\texttt{0}$.

Nous pouvons modifier une valeur par affectation comme $\texttt{d["A"] = 1}$. Et l’instruction $\texttt{d["D"] = 3}$ ne provoque pas d’erreur.

Une nouvelle clé est créée.

Attention : une instruction comme $\texttt{v = d["E"]}$ provoque une erreur car la clé $\texttt{"E"}$ n'existe pas. La méthode $\texttt{get}$ permet de gérer ce genre de problème.

```
v = d.get("A")
v
```

```
v = d.get("E")
print(v)
```

**Nombre d’éléments :**

la fonction $\texttt{len}$ renvoie le nombre d’éléments d’un dictionnaire, sa longueur.

**Suppression d’un élément**

Pour supprimer un élément, nous utilisons l’instruction $\texttt{del d[clé]}$.

```
d = {'A': 0,'B': 1,'C': 2,'D': 3}
del d["D"]
d
```

###<font color="red">Application</font>

Les photos  prises  avec  un  appareil  numérique  contiennent  de  nombreuses  informations.  Dans  le  fichier  image,  par  exemple  au format  jpeg,  sont  stockées  des  données  non  seulement  sur  l’image  elle-même  mais  aussi  sur  l’appareil,  le  logiciel  utilisé,  et  en particulier des données EXIF (Exchangeable Image File Format). Une partie est accessible dans les propriétés de fichier ou avec un logiciel de traitement d’images. Les  spécifications  sont  gérées  par  un  organisme  japonais,  le  JEITA qui  définit  différents  dictionnaires  de  référence  permettant d’accéder  à  ces  données.  Les  clés  (les  tags)  et  les  valeurs  sont  des  nombres  écrits  dans  l’entête  du  fichier.  Les  dictionnaires donnent l’interprétation de ces clés. Par exemple la clé 256 a pour valeur $\texttt{'Width'}$, la largeur de l’image en pixel, la clé 257 a pourvaleur $\texttt{'Length'}$, la hauteur de l’image en pixel.
